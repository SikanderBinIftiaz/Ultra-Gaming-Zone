```python
# ============================================================
# TASK 1: SOLVE SYSTEM OF LINEAR EQUATIONS (AX = B)
# ============================================================

import numpy as np

# First two data points
A = np.array([
    [1, 2],
    [2, 1]
])

B = np.array([
    2.1,
    3.4
])

# Solve AX = B
solution = np.linalg.solve(A, B)

x = solution[0]
y = solution[1]

print("x =", x)
print("y =", y)


# ============================================================
# TASK 2: LOGISTIC REGRESSION USING NUMPY
# ============================================================

# Dataset
time_studied = np.array([1, 2, 4, 5, 2])
cgpa = np.array([2.1, 3.4, 2.8, 4.5, 3.9])

# Use x and y from Task 1 as weights
w1 = x
w2 = y

# Create feature matrix
X = np.column_stack((time_studied, cgpa))

# Linear combination
z = np.dot(X, np.array([w1, w2]))

# Sigmoid function
def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# Probability scores
probabilities = sigmoid(z)

# Apply threshold
predictions = (probabilities >= 0.5).astype(int)

print("\nProbability Scores:")
print(probabilities)

print("\nPredictions:")
print(predictions)


# ============================================================
# TASK 3: LINEAR REGRESSION USING NUMPY
# ============================================================

# Mean values
x_mean = np.mean(time_studied)
y_mean = np.mean(cgpa)

# Calculate slope (b)
b = np.sum((time_studied - x_mean) * (cgpa - y_mean)) / \
    np.sum((time_studied - x_mean) ** 2)

# Calculate intercept (a)
a = y_mean - b * x_mean

# Regression model
y_pred = a + b * time_studied

print("\nIntercept (a) =", a)
print("Slope (b) =", b)

print("\nRegression Equation:")
print(f"y = {a:.4f} + {b:.4f}(X)")

print("\nPredicted Values:")
print(y_pred)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Set random seed for reproducible "diagnostic" noise
np.random.seed(42)

# Dimensions for our mock medical imagery
height, width = 100, 100

# 1. Structural Representations
# Binary Image (e.g., a simple mask)
binary_img = np.random.choice([0, 1], size=(height, width)).astype(np.uint8)

# Grayscale Image (e.g., an X-Ray or Ultrasound slice)
grayscale_img = np.random.randint(0, 256, size=(height, width), dtype=np.uint8)

# RGB Image (e.g., a histological stain sample)
rgb_img = np.random.randint(0, 256, size=(height, width, 3), dtype=np.uint8)

# Print structural dimensions to demonstrate understanding
print("--- Matrix Structural Dimensions ---")
print(f"Binary Image Shape:   {binary_img.shape} (2D)")
print(f"Grayscale Image Shape:{grayscale_img.shape} (2D)")
print(f"RGB Image Shape:      {rgb_img.shape} (3D - Height, Width, Channels)\n")


# 2. Image Modification Filter (Altering a specific slice/ROI)
# Let's say we detect an anomaly in a specific 30x30 slice of the grayscale image.
# We will apply a "contrast boost" filter to this Region of Interest (ROI).
modified_grayscale = grayscale_img.copy()

# Slicing a specific pixel coordinate box: rows 35-65, columns 35-65
row_start, row_end = 35, 65
col_start, col_end = 35, 65

# Apply filter: Maximize brightness within this slice to highlight a "tumor/lesion"
modified_grayscale[row_start:row_end, col_start:col_end] = 255


# Visualization Pipeline
fig, axes = plt.subplots(1, 4, figsize=(18, 5))

# Plot Binary
axes[0].imshow(binary_img, cmap='gray')
axes[0].set_title("Binary Matrix\n(Shape: 100x100)", fontsize=11)
axes[0].axis('off')

# Plot Original Grayscale
axes[1].imshow(grayscale_img, cmap='gray')
axes[1].set_title("Original Grayscale\n(Shape: 100x100)", fontsize=11)
axes[1].axis('off')

# Plot Modified Grayscale (Filtered Slice)
axes[2].imshow(modified_grayscale, cmap='gray')
# Draw a subtle red bounding box around the modified slice for clarity
rect = plt.Rectangle((col_start, row_start), col_end-col_start, row_end-row_start, 
                     linewidth=1.5, edgecolor='red', facecolor='none')
axes[2].add_patch(rect)
axes[2].set_title("Filtered Grayscale\n(Modified 30x30 Slice)", fontsize=11)
axes[2].axis('off')

# Plot RGB
axes[3].imshow(rgb_img)
axes[3].set_title("RGB Matrix\n(Shape: 100x100x3)", fontsize=11)
axes[3].axis('off')

plt.tight_layout()
plt.show()
```

```python
import pandas as pd

#Synthesize a mock DiaTech Heart Disease Tabular Dataset
data = {
    'Patient_ID': [f'P_{i:03d}' for i in range(1, 11)],
    'Age': [45, 62, 38, 71, 58, 67, 52, 41, 69, 50],
    'Blood_Pressure': [120, 145, 115, 160, 138, 150, 132, 118, 155, 128],
    'Cholesterol': [210, 265, 190, 290, 240, 310, 225, 205, 280, 245],
    'Target_Heart_Disease': [0, 1, 0, 1, 1, 1, 0, 0, 1, 1]  # 1 = High Risk, 0 = Low Risk
}
df = pd.DataFrame(data)
df.set_index('Patient_ID', inplace=True)
print("--- Initial Patient Medical Logs ---")
print(df, "\n")


# Isolate High-Risk Records using .loc (Label-based indexing with conditions)
# Risk condition: Blood Pressure > 130 OR Cholesterol > 240
high_risk_condition = (df['Blood_Pressure'] > 130) | (df['Cholesterol'] > 240)
isolated_high_risk = df.loc[high_risk_condition]

print("--- Isolated High-Risk Records (.loc) ---")
print(isolated_high_risk, "\n")


#Combine .loc and .iloc to extract a Targeted Sample Slice
# Suppose we want to extract the first 4 rows of our high-risk subset for manual inspection.
# We use .iloc on the already .loc-filtered dataframe.
targeted_sample = isolated_high_risk.iloc[0:4]

print("--- Targeted Sample Slice (.iloc of first 4 high-risk records) ---")
print(targeted_sample, "\n")


# Apply .groupby() aggregation to evaluate key medical trends
# We group by 'Target_Heart_Disease' status within our sample to see averages of clinical metrics.
trends = targeted_sample.groupby('Target_Heart_Disease').agg({
    'Age': 'mean',
    'Blood_Pressure': 'mean',
    'Cholesterol': ['mean', 'max']
}).round(1)

print("--- Evaluated Medical Trends Within Sample ---")
print(trends)
```

```python
import sqlite3
from datetime import datetime

class UltraCafeManager:
    def __init__(self, db_name="ultra_gaming_zone.db"):
        self.conn = sqlite3.connect(db_name)
        self._initialize_db()

    def _initialize_db(self):
        """Creates tables for PC sessions and general sales."""
        # PC Sessions Table
        self.conn.execute("""
            CREATE TABLE IF NOT EXISTS pc_sessions (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                cabin_no TEXT,
                start_time TEXT,
                end_time TEXT,
                total_charge REAL,
                date_logged DATE DEFAULT (DATE('now'))
            )
        """)
        # Other Sales Table (Snacks, Printing, etc.)
        self.conn.execute("""
            CREATE TABLE IF NOT EXISTS other_sales (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                item_name TEXT,
                amount REAL,
                date_logged DATE DEFAULT (DATE('now'))
            )
        """)
        self.conn.commit()

    # --- 1. CORE CALCULATION LOGIC ---
    def calculate_pc_bill(self, start_str, end_str):
        """Pricing: 40 Rs for <= 30 mins, 70 Rs/hr for > 30 mins."""
        fmt = "%Y-%m-%d %H:%M:%S"
        start = datetime.strptime(start_str, fmt)
        end = datetime.strptime(end_str, fmt)
        
        diff_mins = int((end - start).total_seconds() / 60)
        
        if diff_mins <= 0: return 0
        if diff_mins <= 30:
            return 40.0
        else:
            return round((diff_mins / 60.0) * 70.0, 2)

    # --- 2. RECORDING DATA ---
    def add_pc_session(self, cabin, start, end):
        charge = self.calculate_pc_bill(start, end)
        self.conn.execute(
            "INSERT INTO pc_sessions (cabin_no, start_time, end_time, total_charge) VALUES (?,?,?,?)",
            (cabin, start, end, charge)
        )
        self.conn.commit()
        return charge

    def add_other_sale(self, item, amount):
        self.conn.execute("INSERT INTO other_sales (item_name, amount) VALUES (?,?)", (item, amount))
        self.conn.commit()

    # --- 3. DASHBOARD ANALYTICS ---
    def get_daily_total(self):
        """Total from PC sessions + Other sales for today."""
        today = datetime.now().strftime("%Y-%m-%d")
        
        pc_total = self.conn.execute("SELECT SUM(total_charge) FROM pc_sessions WHERE date_logged = ?", (today,)).fetchone()[0] or 0
        misc_total = self.conn.execute("SELECT SUM(amount) FROM other_sales WHERE date_logged = ?", (today,)).fetchone()[0] or 0
        
        return pc_total + misc_total

    def get_monthly_total(self):
        """Total earnings for the current month."""
        month = datetime.now().strftime("%Y-%m")
        
        pc_total = self.conn.execute("SELECT SUM(total_charge) FROM pc_sessions WHERE date_logged LIKE ?", (f"{month}%",)).fetchone()[0] or 0
        misc_total = self.conn.execute("SELECT SUM(amount) FROM other_sales WHERE date_logged LIKE ?", (f"{month}%",)).fetchone()[0] or 0
        
        return pc_total + misc_total
```
